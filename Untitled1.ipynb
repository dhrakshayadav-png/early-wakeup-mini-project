```python
# ==========================================
# STEP 3: EXPLORATORY DATA ANALYSIS (EDA)
# ==========================================

# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset
df = pd.read_excel("cleaned_makeup_data.xlsx")

# ==========================================
# 1. BASIC INFORMATION
# ==========================================
print("="*50)
print("DATASET SHAPE")
print("="*50)
print(df.shape)

print("\n" + "="*50)
print("DATASET INFO")
print("="*50)
print(df.info())

print("\n" + "="*50)
print("FIRST 5 ROWS")
print("="*50)
print(df.head())

# ==========================================
# 2. MISSING VALUES
# ==========================================
print("\n" + "="*50)
print("MISSING VALUES")
print("="*50)
print(df.isnull().sum())

# ==========================================
# 3. DUPLICATE VALUES
# ==========================================
print("\n" + "="*50)
print("DUPLICATE RECORDS")
print("="*50)
print(df.duplicated().sum())

# ==========================================
# 4. STATISTICAL SUMMARY
# ==========================================
print("\n" + "="*50)
print("STATISTICAL SUMMARY")
print("="*50)
print(df.describe())

# ==========================================
# 5. NUMERICAL & CATEGORICAL COLUMNS
# ==========================================
num_cols = df.select_dtypes(include=np.number).columns
cat_cols = df.select_dtypes(exclude=np.number).columns

print("\nNumerical Columns:")
print(list(num_cols))

print("\nCategorical Columns:")
print(list(cat_cols))

# ==========================================
# 6. HISTOGRAMS FOR NUMERICAL COLUMNS
# ==========================================
for col in num_cols:
    plt.figure(figsize=(6,4))
    sns.histplot(df[col], kde=True)
    plt.title(f"Distribution of {col}")
    plt.tight_layout()
    plt.show()

# ==========================================
# 7. BOXPLOTS FOR OUTLIER DETECTION
# ==========================================
for col in num_cols:
    plt.figure(figsize=(6,4))
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot of {col}")
    plt.tight_layout()
    plt.show()

# ==========================================
# 8. CATEGORICAL COLUMN ANALYSIS
# ==========================================
for col in cat_cols:
    print(f"\nValue Counts for {col}")
    print(df[col].value_counts())

    plt.figure(figsize=(8,4))
    df[col].value_counts().plot(kind='bar')
    plt.title(f"Count Plot - {col}")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

# ==========================================
# 9. CORRELATION HEATMAP
# ==========================================
if len(num_cols) > 1:
    plt.figure(figsize=(14,10))
    corr = df[num_cols].corr()

    sns.heatmap(
        corr,
        annot=True,
        cmap="coolwarm",
        fmt=".2f"
    )

    plt.title("Correlation Heatmap")
    plt.tight_layout()
    plt.show()

# ==========================================
# 10. SCATTER PLOTS
# ==========================================
if len(num_cols) >= 2:

    target = num_cols[-1]  # last numerical column

    for col in num_cols[:-1]:
        plt.figure(figsize=(6,4))
        sns.scatterplot(x=df[col], y=df[target])
        plt.title(f"{col} vs {target}")
        plt.tight_layout()
        plt.show()

# ==========================================
# 11. PAIRPLOT
# ==========================================
if len(num_cols) <= 10:
    sns.pairplot(df[num_cols])
    plt.show()

# ==========================================
# 12. EDA SUMMARY
# ==========================================
print("\n" + "="*50)
print("EDA COMPLETED SUCCESSFULLY")
print("="*50)
print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")
print(f"Missing Values: {df.isnull().sum().sum()}")
print(f"Duplicate Records: {df.duplicated().sum()}")
print("EDA Analysis Finished")
```
